# Workflow widget

This example starts the batch-reduction GUI with three Amor runs: a supermirror reference (1632) and two NiTi multilayer measurements (1634 and 1635) at different sample angles. Both sample measurements are reduced and plotted automatically.

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
%matplotlib widget
```

```python
from pathlib import Path
from shutil import copy2
from tempfile import TemporaryDirectory

from ess.amor import data
from ess.reflectometry.gui import AmorBatchReductionGUI
```

The example files are downloaded on first use and cached. A temporary proposal folder collects the three files for the GUI.

```python
example_directory = TemporaryDirectory(prefix='amor-widget-')
proposal_dir = Path(example_directory.name)

for run in (1632, 1634, 1635):
    filename = data.amor_run(run)
    copy2(filename, proposal_dir / filename.name)
```

Mark run 1632 as the reference, then reduce and plot both sample runs. These steps correspond to checking **Reference** in the runs table and pressing **Reduce** and **Plot**. With no reduction rows selected, both samples are processed.

```python
gui = AmorBatchReductionGUI()
gui.proposal_number_box.value = str(proposal_dir)

runs = gui.runs_table.data
reference_index = runs.index[runs['Run'] == '1632'][0]
gui.runs_table.set_cell_value('Reference', reference_index, True)

gui.run_workflow()
gui.display_results()
gui.widget
```

The documentation displays the populated tables and the resulting curves. Open this notebook with a running kernel to change the Q range, toggle **R*Q^4**, or repeat the reduction. To work with your own data, enter its folder path in **Proposal no.**.
